##This analysis examines the job market in Seoul using job posting data collected and refined from the Saramin website.

##본 분석에서는 사람인 웹사이트에서 수집하고 정제한 채용 공고 데이터를 활용하여 서울의 채용 시장을 분석합니다.

```python
import duckdb

con = duckdb.connect(database="saramin_jobs")

con.execute("""
    SELECT COUNT(*) AS total_jobs
    FROM read_csv_auto('saramin_jobs.csv')
    """).df()
```

```python
con.close()
```

##1. What types of jobs are most in demand? (어떤 직무의 채용 수요가 가장 높은가?)

"Other" means the job posting's sector text didn't contain any of the 5 keywords used in the categorize() function (Data Analysis, Data Engineer, Deep Learning, Machine Learning, Data Science). It likely uses a different but related term (e.g., "Big Data", "AI", "Statistics") that wasn't included in the keyword list.

"Other"는 해당 채용공고의 sector 텍스트에 categorize() 함수에서 사용한 5개 키워드(데이터분석가, 데이터엔지니어, 딥러닝, 머신러닝, 데이터 사이언티스트) 중 어느 것도 포함되지 않았다는 의미다. 목록에 없는 다른 관련 용어(예: "빅데이터", "AI", "통계" 등)를 사용했을 가능성이 높다.

```python
con1 = duckdb.connect(database="saramin_jobs")

con1.execute("""
    SELECT category,
    COUNT(*) AS job_count
    FROM read_csv_auto('saramin_jobs.csv')
    GROUP BY category
    ORDER BY job_count DESC;
    """).df()
```

```python
con1.close()
```

##2. What are the career requirements for these job postings? (채용 공고의 경력 요건은 어떻게 구성되어 있는가?)

```python
con2 = duckdb.connect(database="saramin_jobs")

con2.execute("""
    SELECT career,
    COUNT(*) AS job_count
    FROM read_csv_auto('saramin_jobs.csv')
    GROUP BY career
    ORDER BY job_count DESC;
    """).df()
```

```python
con2.close()
```

##3. Which job categories offer the most opportunities for entry-level candidates? (신입 지원자에게 가장 많은 기회를 제공하는 직무는 무엇인가?)

```python
con3 = duckdb.connect(database="saramin_jobs")

con3.execute("""
    SELECT category, career,
    COUNT(*) AS job_count
    FROM read_csv_auto('saramin_jobs.csv')
    GROUP BY category, career
    ORDER BY category, job_count DESC;
    """).df()
```

```python
con3.close()
```

##4. Which industries have the highest demand for talent? (어떤 산업에서 인재 채용 수요가 가장 높은가?)

```python
con4 = duckdb.connect(database="saramin_jobs")

con4.execute("""
    SELECT sector,
    COUNT(*) AS job_count
    FROM read_csv_auto('saramin_jobs.csv')
    GROUP BY sector
    ORDER BY job_count DESC
    """).df()
```

```python
con4.close()
```

##5. Where are job opportunities concentrated? (채용 기회는 어느 지역에 가장 많이 집중되어 있는가?)

```python
con5 = duckdb.connect(database="saramin_jobs")

con5.execute("""
    SELECT location,
    COUNT(*) AS job_count
    FROM read_csv_auto('saramin_jobs.csv')
    GROUP BY location
    ORDER BY job_count DESC
    """).df()
```

```python
con5.close()
```

##6. What level of education is most commonly required? (가장 많이 요구되는 학력 수준은 무엇인가?)

```python
con6 = duckdb.connect(database="saramin_jobs")

con6.execute("""
    SELECT education,
    COUNT(*) AS job_count
    FROM read_csv_auto('saramin_jobs.csv')
    GROUP BY education
    ORDER BY job_count DESC
    """).df()
```

```python
con6.close()
```
